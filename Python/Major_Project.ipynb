```python
iqr = df['Quantity'].quantile(0.75) - df['Quantity'].quantile(0.25)
upper_threshold = df['Quantity'].quantile(0.75) + (1.5*iqr)  #Q3+1.5*IQR
lower_thrshold = df['Quantity'].quantile(0.25) - (1.5*iqr)   #Q1-1.5*IQR
print(upper_threshold,",",lower_thrshold)
```

```python
df['Quantity']=df['Quantity'].clip(lower_thrshold,upper_threshold)
```

```python
iqr = df['Sales'].quantile(0.75) - df['Sales'].quantile(0.25)
upper_threshold = df['Sales'].quantile(0.75) + (1.5*iqr)  #Q3+1.5*IQR
lower_thrshold = df['Sales'].quantile(0.25) - (1.5*iqr)   #Q1-1.5*IQR
print(upper_threshold,",",lower_thrshold)
```

```python
df['Sales']=df['Sales'].clip(lower_thrshold,upper_threshold)
```

```python
iqr = df['Profit'].quantile(0.75) - df['Profit'].quantile(0.25)
upper_threshold = df['Profit'].quantile(0.75) + (1.5*iqr)  #Q3+1.5*IQR
lower_thrshold = df['Profit'].quantile(0.25) - (1.5*iqr)   #Q1-1.5*IQR
print(upper_threshold,",",lower_thrshold)
```

```python
df['Profit']=df['Profit'].clip(lower_thrshold,upper_threshold)
```
